# BDR-AI Decision Evidence Signals v1 - Quickstart Guide

This notebook demonstrates how to load and explore the **BDR-AI/decision_evidence_signals_v1** dataset.

## Dataset Overview

- **Purpose**: Synthetic advisory evidence signals for insurance and justice domains
- **Records**: 300 synthetic records
- **Governance**: 100% advisory-only, human-in-the-loop required
- **License**: Apache 2.0

## 2. Load Dataset from Hugging Face

Load the dataset directly from Hugging Face:

````python
from datasets import load_dataset
import pandas as pd

# Load dataset
dataset = load_dataset("BDR-AI/decision_evidence_signals_v1")

# Convert to pandas DataFrame
df = dataset['train'].to_pandas()

print(f"Loaded {len(df)} records")
df.head()
````

## 3. Explore Dataset Structure

Examine the schema and data types:

````python
# Display column information
print("Dataset Schema:")
print(df.dtypes)
print(f"\nShape: {df.shape}")
print(f"\nColumns: {list(df.columns)}")
````

## 4. Domain Distribution

Check the balance between insurance and justice domains:

````python
# Domain distribution
domain_counts = df['domain'].value_counts()
print("Domain Distribution:")
print(domain_counts)
print(f"\nPercentages:")
print(domain_counts / len(df) * 100)
````

## 5. Signal Type Analysis

Explore the different types of evidence signals:

````python
# Signal type distribution
signal_counts = df['signal_type'].value_counts()
print("Signal Type Distribution:")
print(signal_counts)

# Signal types by domain
print("\nSignal Types by Domain:")
print(pd.crosstab(df['domain'], df['signal_type']))
````

## 6. Risk Weight Distribution

Analyze risk weight categories:

````python
# Risk weight distribution
risk_counts = df['risk_weight'].value_counts()
print("Risk Weight Distribution:")
print(risk_counts)

# Risk weights by domain
print("\nRisk Weights by Domain:")
print(pd.crosstab(df['domain'], df['risk_weight']))
````

## 7. Filter High-Risk Signals

Example: Filter for high-risk insurance signals:

````python
# Filter high-risk insurance signals
high_risk_insurance = df[
    (df['domain'] == 'insurance') & 
    (df['risk_weight'] == 'high')
]

print(f"Found {len(high_risk_insurance)} high-risk insurance signals")
print("\nSample records:")
high_risk_insurance[['record_id', 'use_case', 'signal_type', 'evidence_text']].head(3)
````

## 8. Examine Evidence Text

View sample evidence narratives:

````python
# Display sample evidence text
sample = df.sample(3, random_state=42)

for idx, row in sample.iterrows():
    print(f"\n{'='*70}")
    print(f"Record ID: {row['record_id']}")
    print(f"Domain: {row['domain']} | Use Case: {row['use_case']}")
    print(f"Signal Type: {row['signal_type']} | Risk: {row['risk_weight']}")
    print(f"\nEvidence:")
    print(row['evidence_text'])
    print(f"\nAdvisory Note:")
    print(row['advisory_note'])
    print(f"\nGovernance: {row['governance_tag']}")
````

## 9. Use Case Analysis

Explore different use cases by domain:

````python
# Use case distribution
print("Insurance Use Cases:")
print(df[df['domain'] == 'insurance']['use_case'].value_counts())

print("\nJustice Use Cases:")
print(df[df['domain'] == 'justice']['use_case'].value_counts())
````

## 10. Governance Verification

Verify all records have proper governance tags:

````python
# Check governance tags
governance_check = df['governance_tag'].value_counts()
print("Governance Tag Distribution:")
print(governance_check)

# Verify 100% compliance
expected_tag = "advisory_only_human_review_required"
compliance_rate = (df['governance_tag'] == expected_tag).sum() / len(df) * 100
print(f"\nGovernance Compliance: {compliance_rate:.1f}%")
````

## 11. Export Filtered Data

Example: Export justice domain signals to CSV:

````python
# Filter and export
justice_signals = df[df['domain'] == 'justice']

# Uncomment to save
# justice_signals.to_csv('justice_signals.csv', index=False)

print(f"Filtered {len(justice_signals)} justice domain signals")
````

## 12. Summary Statistics

Generate comprehensive summary:

````python
print("Dataset Summary:")
print(f"Total Records: {len(df)}")
print(f"\nDomains: {df['domain'].nunique()}")
print(f"Signal Types: {df['signal_type'].nunique()}")
print(f"Use Cases: {df['use_case'].nunique()}")
print(f"Risk Levels: {df['risk_weight'].nunique()}")
print(f"\nDate Range: {df['created_at'].min()} to {df['created_at'].max()}")
print(f"\nGovernance Compliance: 100%")
print(f"Synthetic Data: Yes (No PII)")
````
